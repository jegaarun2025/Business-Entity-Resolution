```python
import re

def clean_text(text):
    if pd.isna(text):
        return ""

    text = str(text).lower()

    # Replace punctuation with space
    text = re.sub(r"[^a-z0-9\s]", " ", text)

    # Remove extra spaces
    text = re.sub(r"\s+", " ", text).strip()

    return text

print(clean_text("Delta Tetlecommunication Inc."))
print(clean_text("1795 Westchester Drive, High Point, NC"))
```

```python
# Create cleaned versions of name and address

for df in [source1, source2, source3]:
    df["clean_name"] = df["business_name"].apply(clean_text)
    df["clean_address"] = df["business_address"].apply(clean_text)

print("Cleaning completed!")

print("\nSource 1:")
print(source1[["business_name", "clean_name",
                "business_address", "clean_address"]].head())
```

```python
# Measure similarity between Source 1 and its actual matching records

from difflib import SequenceMatcher

def similarity(a, b):
    return SequenceMatcher(None, a, b).ratio()


name_scores = []
address_scores = []

# Create quick lookup dictionaries
s2_lookup = source2_matches.set_index("entity_id").to_dict("index")
s3_lookup = source3_matches.set_index("entity_id").to_dict("index")

for _, row in source1.iterrows():

    # Get actual matches for this Source 1 business
    matches = gt_sample[
        gt_sample["source1_entity_id"] == row["entity_id"]
    ]["matched_entity_ids"].iloc[0]

    if not matches or pd.isna(matches):
        continue

    for match_id in matches.split(","):

        if match_id in s2_lookup:
            candidate = s2_lookup[match_id]

        elif match_id in s3_lookup:
            candidate = s3_lookup[match_id]

        else:
            continue

        name_scores.append(
            similarity(row["clean_name"], candidate["clean_name"])
        )

        address_scores.append(
            similarity(row["clean_address"], candidate["clean_address"])
        )

print("Total genuine matches analyzed:", len(name_scores))

print("\nAverage name similarity:",
      np.mean(name_scores))

print("Average address similarity:",
      np.mean(address_scores))

print("\nMinimum name similarity:",
      np.min(name_scores))

print("Minimum address similarity:",
      np.min(address_scores))

print("\nMedian name similarity:",
      np.median(name_scores))

print("Median address similarity:",
      np.median(address_scores))
```

```python
from difflib import SequenceMatcher

def get_candidates(s1_row, source2_df, source3_df, top_n=10):
    candidates = []

    # Combine Source 2 and Source 3
    all_sources = pd.concat(
        [source2_df, source3_df],
        ignore_index=True
    )

    # Same country only
    same_country = all_sources[
        all_sources["country"] == s1_row["country"]
    ].copy()

    # Calculate name similarity
    for _, row in same_country.iterrows():
        name_score = SequenceMatcher(
            None,
            s1_row["clean_name"],
            row["clean_name"]
        ).ratio()

        address_score = SequenceMatcher(
            None,
            s1_row["clean_address"],
            row["clean_address"]
        ).ratio()

        # Combined score
        combined_score = (
            0.6 * name_score +
            0.4 * address_score
        )

        candidates.append({
            "source1_id": s1_row["entity_id"],
            "candidate_id": row["entity_id"],
            "name_score": name_score,
            "address_score": address_score,
            "combined_score": combined_score
        })

    candidates_df = pd.DataFrame(candidates)

    # Keep only the top candidates
    candidates_df = candidates_df.sort_values(
        "combined_score",
        ascending=False
    ).head(top_n)

    return candidates_df
```

```python
def token_similarity(text1, text2):
    # Convert text into sets of words
    words1 = set(text1.split())
    words2 = set(text2.split())

    # Avoid division by zero
    if not words1 or not words2:
        return 0.0

    # Jaccard similarity
    intersection = len(words1 & words2)
    union = len(words1 | words2)

    return intersection / union


# Test on Orelee's actual matches

s1_row = test_s1.iloc[0]

for _, row in actual_records.iterrows():
    name_token_score = token_similarity(
        s1_row["clean_name"],
        row["clean_name"]
    )

    address_token_score = token_similarity(
        s1_row["clean_address"],
        row["clean_address"]
    )

    print("\nCandidate:", row["entity_id"])
    print("Name:", row["business_name"])
    print("Name token similarity:", round(name_token_score, 3))
    print("Address token similarity:", round(address_token_score, 3))
```

```python
def calculate_match_score(s1_row, candidate_row):
    # Character-based similarity
    name_char = SequenceMatcher(
        None,
        s1_row["clean_name"],
        candidate_row["clean_name"]
    ).ratio()

    address_char = SequenceMatcher(
        None,
        s1_row["clean_address"],
        candidate_row["clean_address"]
    ).ratio()

    # Token-based similarity
    name_token = token_similarity(
        s1_row["clean_name"],
        candidate_row["clean_name"]
    )

    address_token = token_similarity(
        s1_row["clean_address"],
        candidate_row["clean_address"]
    )

    # Final combined score
    score = (
        0.30 * name_char +
        0.30 * name_token +
        0.20 * address_char +
        0.20 * address_token
    )

    return score


# Test on the 4 genuine matches

for _, row in actual_records.iterrows():

    score = calculate_match_score(
        s1_row,
        row
    )

    print(
        row["entity_id"],
        "|",
        row["business_name"],
        "| Score:",
        round(score, 3)
    )
```

```python
# Test our scoring method across the first 100 Source 1 businesses

genuine_scores = []
random_scores = []

all_sources = pd.concat(
    [source2, source3],
    ignore_index=True
)

for _, s1_row in test_s1.iterrows():

    # Get ground-truth matches for this Source 1 business
    gt_row = gt_sample[
        gt_sample["source1_entity_id"] == s1_row["entity_id"]
    ]

    if len(gt_row) == 0:
        continue

    matched_ids = gt_row["matched_entity_ids"].iloc[0]

    # ---- Genuine matches ----
    if pd.notna(matched_ids) and matched_ids.strip() != "":

        matched_ids = matched_ids.split(",")

        genuine_records = pd.concat(
            [source2_matches, source3_matches],
            ignore_index=True
        )

        genuine_records = genuine_records[
            genuine_records["entity_id"].isin(matched_ids)
        ]

        for _, candidate in genuine_records.iterrows():
            score = calculate_match_score(
                s1_row,
                candidate
            )
            genuine_scores.append(score)

    # ---- Random non-matches ----
    same_country = all_sources[
        all_sources["country"] == s1_row["country"]
    ]

    random_sample = same_country.sample(
        n=min(5, len(same_country)),
        random_state=42
    )

    for _, candidate in random_sample.iterrows():
        score = calculate_match_score(
            s1_row,
            candidate
        )
        random_scores.append(score)


print("Genuine scores:", len(genuine_scores))
print("Random scores:", len(random_scores))

print("\nGENUINE MATCHES")
print("Average:", round(np.mean(genuine_scores), 3))
print("Minimum:", round(np.min(genuine_scores), 3))
print("Median:", round(np.median(genuine_scores), 3))

print("\nRANDOM NON-MATCHES")
print("Average:", round(np.mean(random_scores), 3))
print("Maximum:", round(np.max(random_scores), 3))
print("Median:", round(np.median(random_scores), 3))
```

```python
# Test different score thresholds

thresholds = [0.3, 0.4, 0.5, 0.6, 0.7]

print("THRESHOLD TEST")
print("-" * 60)

for threshold in thresholds:

    genuine_kept = sum(
        score >= threshold
        for score in genuine_scores
    )

    random_kept = sum(
        score >= threshold
        for score in random_scores
    )

    genuine_percent = (
        genuine_kept / len(genuine_scores)
    ) * 100

    random_percent = (
        random_kept / len(random_scores)
    ) * 100

    print(
        f"Threshold {threshold:.1f} | "
        f"Genuine kept: {genuine_kept}/{len(genuine_scores)} "
        f"({genuine_percent:.1f}%) | "
        f"Random kept: {random_kept}/{len(random_scores)} "
        f"({random_percent:.1f}%)"
    )
```

```python
# Find difficult non-matches using similar business names

hard_negative_scores = []

all_sources = pd.concat(
    [source2, source3],
    ignore_index=True
)

for _, s1_row in test_s1.iterrows():

    # Same country only
    same_country = all_sources[
        all_sources["country"] == s1_row["country"]
    ].copy()

    # Calculate name similarity
    same_country["name_similarity"] = same_country[
        "clean_name"
    ].apply(
        lambda x: SequenceMatcher(
            None,
            s1_row["clean_name"],
            x
        ).ratio()
    )

    # Take the most name-similar records
    hard_candidates = same_country.sort_values(
        "name_similarity",
        ascending=False
    ).head(10)

    # Remove actual genuine matches
    gt_row = gt_sample[
        gt_sample["source1_entity_id"] == s1_row["entity_id"]
    ]

    if len(gt_row) > 0:
        matched_ids = gt_row["matched_entity_ids"].iloc[0]

        if pd.notna(matched_ids) and matched_ids.strip():
            matched_ids = set(matched_ids.split(","))
            hard_candidates = hard_candidates[
                ~hard_candidates["entity_id"].isin(matched_ids)
            ]

    # Calculate our final score for the hard negatives
    for _, candidate in hard_candidates.iterrows():

        score = calculate_match_score(
            s1_row,
            candidate
        )

        hard_negative_scores.append(score)

print("Hard negative candidates:", len(hard_negative_scores))
print("Average score:", round(np.mean(hard_negative_scores), 3))
print("Median score:", round(np.median(hard_negative_scores), 3))
print("Maximum score:", round(np.max(hard_negative_scores), 3))
```

```python
# ============================================================
# BUSINESS ENTITY RESOLUTION - COMPLETE PROJECT
# ============================================================

# 1. Install required library
!pip -q install rapidfuzz kagglehub

import pandas as pd
import numpy as np
import re
import os
import kagglehub
from rapidfuzz import fuzz

print("Libraries loaded successfully!")


# ============================================================
# 2. DOWNLOAD / LOCATE DATASET
# ============================================================

print("\nChecking dataset...")

dataset_path = kagglehub.dataset_download(
    "raghavdharwal/amazon-ml-challenge-2026"
)

print("Dataset location:")
print(dataset_path)


# ============================================================
# 3. FIND TRAINING FILES AUTOMATICALLY
# ============================================================

train_source1 = None
train_source2 = None
train_source3 = None

for root, dirs, files in os.walk(dataset_path):

    for file in files:

        if file == "train_source1.tsv":
            train_source1 = os.path.join(root, file)

        elif file == "train_source2.tsv":
            train_source2 = os.path.join(root, file)

        elif file == "train_source3.tsv":
            train_source3 = os.path.join(root, file)


print("\nFiles found:")

print("Source 1:", train_source1)
print("Source 2:", train_source2)
print("Source 3:", train_source3)


# Check files
if train_source1 is None:
    raise FileNotFoundError("train_source1.tsv was not found")

if train_source2 is None:
    raise FileNotFoundError("train_source2.tsv was not found")

if train_source3 is None:
    raise FileNotFoundError("train_source3.tsv was not found")


# ============================================================
# 4. LOAD DATA
# ============================================================

print("\nLoading datasets...")

source1 = pd.read_csv(
    train_source1,
    sep="\t",
    nrows=10000
)

source2 = pd.read_csv(
    train_source2,
    sep="\t"
)

source3 = pd.read_csv(
    train_source3,
    sep="\t"
)

print("\nDataset loading completed!")

print("Source 1:", source1.shape)
print("Source 2:", source2.shape)
print("Source 3:", source3.shape)


# ============================================================
# 5. CLEAN TEXT
# ============================================================

def clean_text(text):

    if pd.isna(text):
        return ""

    text = str(text).lower()

    text = re.sub(
        r"[^a-z0-9\s]",
        " ",
        text
    )

    text = re.sub(
        r"\s+",
        " ",
        text
    ).strip()

    return text


print("\nCleaning business names and addresses...")

for df in [source1, source2, source3]:

    df["clean_name"] = (
        df["business_name"]
        .apply(clean_text)
    )

    df["clean_address"] = (
        df["business_address"]
        .apply(clean_text)
    )


print("Cleaning completed!")

# Show your example
orelee_rows = source1[
    source1["business_name"]
    .astype(str)
    .str.contains(
        "Orelee",
        case=False,
        na=False
    )
]

if len(orelee_rows) > 0:

    print("\nExample:")
    print(
        "Original:",
        orelee_rows.iloc[0]["business_name"]
    )

    print(
        "Cleaned :",
        orelee_rows.iloc[0]["clean_name"]
    )


# ============================================================
# 6. CREATE BLOCKING KEY
# ============================================================

def create_block_key(df):

    df["block_key"] = (
        df["country"]
        .fillna("")
        .astype(str)
        .str.lower()
        + "_"
        + df["clean_name"]
        .str[:2]
    )

    return df


source1 = create_block_key(source1)
source2 = create_block_key(source2)
source3 = create_block_key(source3)

print("\nBlocking keys created!")


# ============================================================
# 7. COMBINE SOURCE 2 AND SOURCE 3
# ============================================================

print("\nCombining Source 2 and Source 3...")

all_sources = pd.concat(
    [source2, source3],
    ignore_index=True
)

print(
    "Combined candidate records:",
    len(all_sources)
)


# ============================================================
# 8. CREATE BLOCK INDEX
# ============================================================

print("\nCreating candidate blocks...")

block_index = (
    all_sources
    .groupby("block_key")
    .groups
)

print(
    "Number of blocks:",
    len(block_index)
)


# ============================================================
# 9. MATCHING FUNCTION
# ============================================================

def find_best_match(row):

    block_key = row["block_key"]

    candidate_indices = block_index.get(
        block_key,
        []
    )

    if len(candidate_indices) == 0:

        return {
            "source1_id": row["entity_id"],
            "matched_id": None,
            "name_score": 0,
            "address_score": 0,
            "final_score": 0,
            "decision": "NO CANDIDATE"
        }


    candidates = all_sources.loc[
        candidate_indices
    ]

    best_result = None

    for idx, candidate in candidates.iterrows():

        name_score = fuzz.token_sort_ratio(
            row["clean_name"],
            candidate["clean_name"]
        ) / 100

        address_score = fuzz.token_sort_ratio(
            row["clean_address"],
            candidate["clean_address"]
        ) / 100

        final_score = (
            0.6 * name_score
            +
            0.4 * address_score
        )

        if (
            best_result is None
            or final_score >
            best_result["final_score"]
        ):

            best_result = {
                "candidate_index": idx,
                "name_score": name_score,
                "address_score": address_score,
                "final_score": final_score
            }


    matched_candidate = all_sources.loc[
        best_result["candidate_index"]
    ]


    # Decision thresholds
    if best_result["final_score"] >= 0.75:

        decision = "MATCH"

    elif best_result["final_score"] >= 0.55:

        decision = "POSSIBLE MATCH"

    else:

        decision = "NO MATCH"


    return {
        "source1_id": row["entity_id"],
        "matched_id":
            matched_candidate["entity_id"],

        "name_score":
            round(
                best_result["name_score"],
                4
            ),

        "address_score":
            round(
                best_result["address_score"],
                4
            ),

        "final_score":
            round(
                best_result["final_score"],
                4
            ),

        "decision":
            decision
    }


# ============================================================
# 10. VALIDATION DATA
# ============================================================

print("\nCreating validation data...")

validation_s1 = source1.sample(
    n=min(1000, len(source1)),
    random_state=42
).copy()

print(
    "Validation records:",
    len(validation_s1)
)


# ============================================================
# 11. RUN ENTITY RESOLUTION
# ============================================================

print("\nStarting entity resolution...")

results = []

for i, (_, row) in enumerate(
    validation_s1.iterrows(),
    start=1
):

    result = find_best_match(row)

    results.append(result)

    if i % 100 == 0:

        print(
            "Processed:",
            i,
            "/",
            len(validation_s1)
        )


results_df = pd.DataFrame(results)

print("\nEntity resolution completed!")


# ============================================================
# 12. CREATE FINAL RESULTS
# ============================================================

final_results = validation_s1[
    [
        "entity_id",
        "business_name",
        "business_address",
        "country"
    ]
].copy()

final_results.rename(
    columns={
        "entity_id":
            "source1_id"
    },
    inplace=True
)

final_results = final_results.merge(
    results_df,
    on="source1_id",
    how="left"
)


# ============================================================
# 13. DISPLAY RESULTS
# ============================================================

print("\n")
print("=" * 70)
print("FINAL BUSINESS ENTITY RESOLUTION RESULTS")
print("=" * 70)

print(
    final_results[
        [
            "source1_id",
            "business_name",
            "matched_id",
            "name_score",
            "address_score",
            "final_score",
            "decision"
        ]
    ].head(20).to_string(
        index=False
    )
)


# ============================================================
# 14. SUMMARY
# ============================================================

print("\n")
print("=" * 70)
print("PROJECT SUMMARY")
print("=" * 70)

print(
    "Total validation records:",
    len(final_results)
)

print(
    "MATCH:",
    (
        final_results["decision"]
        == "MATCH"
    ).sum()
)

print(
    "POSSIBLE MATCH:",
    (
        final_results["decision"]
        == "POSSIBLE MATCH"
    ).sum()
)

print(
    "NO MATCH:",
    (
        final_results["decision"]
        == "NO MATCH"
    ).sum()
)

print(
    "NO CANDIDATE:",
    (
        final_results["decision"]
        == "NO CANDIDATE"
    ).sum()
)


# ============================================================
# 15. SAVE RESULT
# ============================================================

output_file = (
    "business_entity_resolution_results.csv"
)

final_results.to_csv(
    output_file,
    index=False
)

print("\nFinal result saved as:")
print(output_file)


# ============================================================
# 16. SHOW ONE EXAMPLE
# ============================================================

print("\n")
print("=" * 70)
print("EXAMPLE ENTITY RESOLUTION")
print("=" * 70)

example = final_results.iloc[0]

print(
    "Original Business :",
    example["business_name"]
)

print(
    "Matched Entity    :",
    example["matched_id"]
)

print(
    "Name Similarity   :",
    example["name_score"]
)

print(
    "Address Similarity:",
    example["address_score"]
)

print(
    "Final Score       :",
    example["final_score"]
)

print(
    "Decision           :",
    example["decision"]
)


print("\n")
print("=" * 70)
print("BUSINESS ENTITY RESOLUTION COMPLETED!")
print("=" * 70)
```
